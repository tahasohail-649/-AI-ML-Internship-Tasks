```python
# Import libraries and load dataset
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Iris dataset
df = sns.load_dataset('iris')

# Display first few rows
print("✅ Data loaded successfully!")
print("\n--- First 5 rows ---")
print(df.head())

# Check shape
print(f"\n📊 Shape: {df.shape}")
```

```python
# Dataset information and statistics
print("\n--- Dataset Info ---")
print(df.info())

print("\n--- Statistical Summary ---")
print(df.describe())

# Check for missing values
print(f"\n🔍 Missing values: {df.isnull().sum().sum()}")
```

```python
# Pairplot - shows relationships between all features
import warnings
warnings.filterwarnings('ignore')

plt.figure(figsize=(10, 8))
sns.pairplot(df, hue='species', diag_kind='hist')
plt.suptitle('Iris Dataset - Feature Relationships', y=1.02, fontsize=14)
plt.show()
```

```python
# Histograms - shows distribution of each feature
df.hist(figsize=(12, 10), bins=20, edgecolor='black', color='skyblue')
plt.suptitle('Distribution of Iris Features', y=1.02, fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# Box plots - identify outliers in each feature by species
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
features = df.columns[:-1]  # All except 'species'

for i, feature in enumerate(features):
    row, col = i // 2, i % 2
    sns.boxplot(x='species', y=feature, data=df, ax=axes[row, col])
    axes[row, col].set_title(f'Box Plot: {feature} by Species', fontsize=12)

plt.suptitle('Outlier Detection in Iris Dataset', y=1.02, fontsize=14)
plt.tight_layout()
plt.show()
```

# 📊 Observations and Insights - Iris Dataset

## 1. Dataset Overview
- **Total Samples:** 150
- **Features:** 4 (Sepal Length, Sepal Width, Petal Length, Petal Width)
- **Target Classes:** 3 (Setosa, Versicolor, Virginica)
- **Missing Values:** None (Data is clean)

## 2. Statistical Summary
| Feature | Mean | Std Dev | Min | Max |
|---------|------|---------|-----|-----|
| Sepal Length | 5.84 | 0.83 | 4.3 | 7.9 |
| Sepal Width | 3.06 | 0.44 | 2.0 | 4.4 |
| Petal Length | 3.76 | 1.77 | 1.0 | 6.9 |
| Petal Width | 1.20 | 0.76 | 0.1 | 2.5 |

## 3. Key Findings from Visualizations

### 📌 Pairplot Analysis:
- **Setosa** species is clearly linearly separable from others
- **Petal Length** and **Petal Width** show strong positive correlation (r ≈ 0.96)
- Versicolor and Virginica have some overlap in features

### 📌 Histogram Analysis:
- Sepal Length: Approximately normal distribution
- Petal Length: Shows 3 distinct peaks (one for each species)
- Sepal Width: Relatively uniform with slight left skew

### 📌 Box Plot Analysis:
- **No extreme outliers** present in any feature
- Setosa has the **smallest** feature values overall
- Virginica has the **largest** petal measurements
- Sepal Width is the most overlapping feature between species

## 4. Conclusion
The Iris dataset is well-structured with clear patterns. Setosa can be easily distinguished from the other two species, while Versicolor and Virginica require more careful analysis using multiple features combined.
